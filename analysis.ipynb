# Analysing instances where running times are better than us

```python
abhsvi_rows = get_all_abhsvi_rs_lines() + get_all_f1_lines(False)
our_rows = get_all_our_rs_lines() + get_all_f1_lines(True)

assert(len(abhsvi_rows) == len(our_rows))
```

```python
print("ABHSVI is better in the following tests:\n")
time_error = ["timeout", "error"]

for (row_a, row_o) in zip(abhsvi_rows, our_rows):
    assert(row_a.benchmark.split(".")[0] == row_o.benchmark.split(".")[0])
    assert(row_a.initial_states == row_o.initial_states)
    assert(row_a.horizon == row_o.horizon)

    if (row_o.time in time_error) and (row_a.time not in time_error):
        print(row_a.benchmark, row_a.horizon, row_a.time, row_o.time)
    elif (row_a.time not in time_error) and (row_o not in time_error):
        if row_a.time < row_o.time:
            print(row_a.benchmark, row_a.horizon, row_a.time, row_o.time)
```

```python
total = 0.0
count = 0
for (row_a, row_o) in zip(abhsvi_rows, our_rows):
    assert(row_a.benchmark.split(".")[0] == row_o.benchmark.split(".")[0])
    assert(row_a.initial_states == row_o.initial_states)
    assert(row_a.horizon == row_o.horizon)

    if (row_o.time in time_error) and (row_a.time not in time_error):
        print(row_a.benchmark, row_a.horizon, row_a.time, row_o.time)
    elif (row_a.time not in time_error) and (row_o not in time_error):
        if row_a.time > row_o.time:
            assert(isinstance(row_a.time, float) and isinstance(row_o.time, float))
            total += row_a.time/row_o.time
            count += 1
print(f"when we are better we are on average {total/count} times faster")
```
