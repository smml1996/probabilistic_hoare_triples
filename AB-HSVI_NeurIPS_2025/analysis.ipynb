```python
gap_index = 3
time_index = 4
```

# Insights of ABHSVI benchmark --these are the POMDP models found in their artifact. https://zenodo.org/records/17425571

```python
tests = [
    "RockSample_POMDP_N3_G1_K2_R18",
    "RockSample_POMDP_N3_G1_K3_R13",
    "RockSample_POMDP_N3_G1_K4_R83",
    "RockSample_POMDP_N3_G2_K3_R4",
    "RockSample_POMDP_N4_G1_K2_R33",
    "RockSample_POMDP_N5_G1_K2_R96",
    "RockSample_POMDP_N6_G1_K2_R77"
]
```

```python
test_to_gaps = {}

for test in tests:
    for horizon in range(1, 8):
        name = f"{test}_{horizon}"
        f = open(os.path.join("Results", f"{name}.txt"), "r")
        lines = f.readlines()

        for line in lines:
            if len(line) == 1:
                break
            tokens = line.split('\t')
            assert(len(tokens) == 5)

            gap = float(tokens[gap_index])
            time = float(tokens[time_index])
            if test_to_gaps.get(name) is None:
                test_to_gaps[name] = [set(), []]
            if gap not in test_to_gaps[name][0]:
                test_to_gaps[name][0].add(gap)
                test_to_gaps[name][1].append((gap, time))
        f.close()

for (key, v) in test_to_gaps.items():
    if len(v[0]) > 1:
        print(key, v[0], v[1])
```

```python
# code to compare the gaps at time 0 and the gap when the test has already timeout
timeouts = []

f = open(os.path.join("my_results", "results_abhsvi.csv"), "r")

test_name_index = 0
horizon_index = 1
time_index = 2


for line in f.readlines()[1:]:
    tokens = line.split(',')
    test_name = tokens[test_name_index]
    horizon = int(tokens[horizon_index])
    time = float(tokens[time_index])

    if time >= 3600:
        timeouts.append(f"{test_name}_{horizon}")


f.close()

for test in timeouts:
    v = test_to_gaps[test]
    min_gap = min(v[0])
    max_gap = max(v[0])

    min_time = min([e[1] for e in v[1] if e[0] == min_gap])
    max_time = max([e[1] for e in v[1] if e[0] == max_gap])
    print(test, min_gap, min_time, max_gap, max_time)
    print()
```

The gap at time 0 is the smallest found. It does not converge fast enough
